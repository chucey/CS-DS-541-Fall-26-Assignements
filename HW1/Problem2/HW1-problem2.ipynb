 ## a)Implementation of Forward Propagation Functions

1) Relu(z):This function takes an input vector and, for each of its elements, returns the maximum
between the element and zero.


```python
import numpy as np
```

```python
'''
this function computes the relu activation function
parameters: z (input nd array)
returns:output nd array after applying relu.

'''
def ReLU(z):
    return np.maximum(0,z)

```

2.forwardPropA(W1, b1, w2, b2, x): This function should compute the forward pass for a single
input vector x using left multiplication convention for weights, i.e. the same used in our class.

```python
'''
this function computes the forward prop for a single input vector x using left multiplication for weights
parameters: 
    W1 is weight of first layer, shape=(n_hidden, n_input)
    w2 is weight of second layer,shape=(n_output, n_hidden)
    b1 is the bias of first layer,shape=(n_hidden,1)
    b2 is the bias of second layer,shape=(n_output,1)
    x is the single input vector, shape=(n_input,1)
returns: output array after doing forward prop
'''

def forwardPropA(W1, b1, w2, b2, x): 


    z1=W1@x+b1
    #(n_hidden,1)=(n_hidden, n_input) @ (n_input,1) + (n_hidden,1)
    h1=ReLU(z1)
    #(n_hidden,1)
    
    z2=w2@h1+b2
    #(n_output,1)=(n_output, n_hidden) @ (n_hidden,1) + (n_output,1)
    h2=ReLU(z2)
    #(n_output,1)
    
    return h2
```

3)forwardPropB(W1, b1, w2, b2, X): Extend forwardPropA to handle a dataset X. This func-
tion should also use left multiplication.

```python
'''
this function computes the forward prop for a dataset X using left multiplication for weights
parameters: 
    W1 is weight of first layer, shape=(n_hidden, n_input)
    w2 is weight of second layer,shape=(n_output, n_hidden)
    b1 is the bias of first layer,shape=(n_hidden,1)
    b2 is the bias of second layer,shape=(n_output,1)
    X is the dataset, shape=(n_input,m)
returns: output array after doing forward prop
'''

def forwardPropB(W1, b1, w2, b2, X): 


    z1=W1@X+b1
    #(n_hidden,m)=(n_hidden, n_input) @ (n_input,m) + (n_hidden,1)
    h1=ReLU(z1)
    #(n_hidden,m)

    z2=w2@h1+b2
    #(n_output,m)=(n_output, n_hidden) @ (n_hidden,m) + (n_output,1)
    h2=ReLU(z2)
    #(n_output,m)
    
    return h2
```

4.forwardPropC(W1, b1, w2, b2, x): This function should compute the forward pass for a single
input vector x using right multiplication convention for weight.

```python
'''
this function computes the forward prop for a single input vector x using right multiplication for weights
parameters: 
    W1 is weight of first layer, shape=(n_input, n_hidden)
    w2 is weight of second layer,shape=(n_hidden,n_output)
    b1 is the bias of first layer,shape=(1,n_hidden)
    b2 is the bias of second layer,shape=(1,n_output)
    x is the single input vector, shape=(1,n_input)
returns: output array after doing forward prop
'''

def forwardPropC(W1, b1, w2, b2, x): 


    z1=x@W1+b1
    #(1,n_hidden)= (1,n_input) @ (n_input, n_hidden) + (1,n_hidden)
    h1=ReLU(z1)
    #(1,n_hidden)
    
    z2=h1@w2+b2
    #(1,n_output)= (1,n_hidden) @ (n_hidden, n_output) + (1,n_output)
    h2=ReLU(z2)
    return h2
    #(1,n_output)
```

5. forwardPropD(W1, b1, w2, b2, X): Extend forwardPropC to handle a dataset X. This func-
tion should also use right multiplication.

```python
'''
this function computes the forward prop for a dataset X using right multiplication for weights
parameters: 
    W1 is weight of first layer, shape=(n_input, n_hidden)
    w2 is weight of second layer,shape=(n_hidden,n_output)
    b1 is the bias of first layer,shape=(1,n_hidden)
    b2 is the bias of second layer,shape=(1,n_output)
    X is the dataset, shape=(m,n_input)
returns: output array after doing forward prop
'''

def forwardPropD(W1, b1, w2, b2, X): 


    z1=X@W1+b1
    #(m,n_hidden)= (m,n_input) @ (n_input, n_hidden) + (1,n_hidden)
    h1=ReLU(z1)
    #(m,n_hidden)
    
    z2=h1@w2+b2
    #(m,n_output)= (m,n_hidden) @ (n_hidden, n_output) + (1,n_output)
    h2=ReLU(z2)
    return h2
    #(m,n_output)
```
