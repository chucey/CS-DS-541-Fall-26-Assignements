## Problem 1: Understanding Numpy Broadcasting

a) Sum (+) with a Column Vector (2 points)

• Create a 2D array A with a shape of (4, 3).

• Create a 2D array B with a shape of (3, 1) using np.array([[0],[1],[2]]).

• Sum A and B using the + operator.

• What is the shape of the resulting array? Hint: This will produce an error. Explain why the broadcasting fails in this case.

```python
import numpy as np
```

## a)

```python
A=np.random.randn(4,3)
print(A)
B=np.array([[0],[1],[2]])
print(B)
```

```python
Qa=A+B
```

This part produce an error as the following:

---------------------------------------------------------------------------

ValueError                                Traceback (most recent call last)

----> 1 Qa=A+B


ValueError: operands could not be broadcast together with shapes (4,3) (3,1) 


#read this documentation: https://numpy.org/doc/stable/user/basics.broadcasting.html (Two dimensions are compatible when they are equal, or one of them is 1)

This sum will produce a error and broadcasting fails in this condition as the shape of A is (4,3) and B is (3,1). 

(4,3)

(3,1)

If we start with the right most dimensions 3 and 1 --> compatible as one of them is 1

The left dimension is 4 and 3 --> they are not equal and one of them is not 1

So one of the pairs fail--> broadcasting not possible.

## b)

b) Sum (+) with a Row Vector (2 points)

• Create a 2D array A with a shape of (4, 3).

• Create a 2D array C with a shape of (1, 3) using np.array([[0, 1, 2]]).

• Sum A and C using the + operator.

• What is the shape of the resulting array? How to interpret the result?


```python
A=np.random.randn(4,3)
print(A)
C=np.array([[0, 1, 2]])
print(C)
```

```python
Qb=A+C
print(Qb)
print(np.shape(Qb))
```

Shape is --> (4,3)

The broadcasting is successful as we have a (4,3) array being added to a (1,3) array.

(4,3)

(1,3)

comparing from right, 3 and 3 are equal --> compatible

left dimensions--> one of them is 1 --> compatible (dimension 1 can be expanded to 4)

c) Sum (+) with a 1D Array (2 points)

• Create a 2D array A with a shape of (3, 3).

• Create a 1D array D using np.array([0, 1, 2]).

• Sum A by D using the + operator.

• What is the shape of the resulting array? How does broadcast allows the sum between

these arrays, given that they originally have different shapes?


```python
A=np.random.randn(3,3)
print(A)
D=np.array([0, 1, 2])
print(D)
print(np.shape(D))
```

```python
Qc=A+D
print(Qc)
print(np.shape(Qc))
```

(3,3)

(3, )    (1-dim and 3 elements)
